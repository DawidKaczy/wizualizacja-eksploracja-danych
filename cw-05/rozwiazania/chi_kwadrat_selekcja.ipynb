# Lab — Wybór zmiennych testem Chi-kwadrat

Zadanie: selekcja cech testem niezależności χ² dla **trzech** baz środowiskowych.

Próg istotności: **α = 0.25** (specyfika danych środowiskowych — luźniejszy niż klasyczne 0.05).

Tok (jak w slajdach): wczytanie → `train`/`test` (30%) → EDA (info, describe, korelacja, rozkłady) → klasa `ChiSquare` → lista ważnych cech → outliery IQR i usunięcie wierszy odstających w ≥2 kolumnach.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline

from sklearn.model_selection import train_test_split
from scipy import stats
from scipy.stats import chi2_contingency
import itertools
import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style='whitegrid')
np.random.seed(42)
DATA = '../zadania/'
```

## Klasa ChiSquare

Hipoteza zerowa: cecha **nie** ma związku z targetem.  
Jeśli `p < α` → odrzucamy H₀ → cecha **IMPORTANT** dla predykcji.

```python
class ChiSquare:
    def __init__(self, dataframe):
        self.df = dataframe
        self.p = None
        self.chi2 = None
        self.dof = None
        self.dfObserved = None
        self.dfExpected = None

    def _print_chisquare_result(self, colX, alpha):
        result = ''
        if self.p < alpha:
            result = f'{colX} is IMPORTANT for Prediction'
        else:
            result = f'{colX} is NOT an important predictor. (Discard {colX} from model)'
        print(result)
        return self.p < alpha

    def TestIndependence(self, colX, colY, alpha=0.25):
        X = self.df[colX].astype(str)
        Y = self.df[colY].astype(str)
        self.dfObserved = pd.crosstab(Y, X)
        self.chi2, self.p, self.dof, expected = stats.chi2_contingency(self.dfObserved.values)
        self.dfExpected = pd.DataFrame(
            expected,
            columns=self.dfObserved.columns,
            index=self.dfObserved.index
        )
        return self._print_chisquare_result(colX, alpha)


def run_chi_selection(dftrain, target, alpha=0.25):
    cT = ChiSquare(dftrain)
    test_columns = [c for c in dftrain.columns if c != target]
    important, discarded = [], []
    rows = []
    for var in test_columns:
        # powtórz test, zbierając p i chi2
        X = dftrain[var].astype(str)
        Y = dftrain[target].astype(str)
        observed = pd.crosstab(Y, X)
        chi2, p, dof, _ = stats.chi2_contingency(observed.values)
        ok = cT.TestIndependence(colX=var, colY=target, alpha=alpha)
        rows.append({'cecha': var, 'chi2': chi2, 'p': p, 'dof': dof, 'wazna': ok})
        (important if ok else discarded).append(var)
    ranking = pd.DataFrame(rows).sort_values('p')
    return important, discarded, ranking


def iqr_outlier_cleanup(dftrain):
    outlier_list = []
    for i in dftrain.columns:
        q1 = np.percentile(dftrain[i], 25)
        q3 = np.percentile(dftrain[i], 75)
        step = 1.5 * (q3 - q1)
        list_out = dftrain[
            (dftrain[i] < q1 - step) | (dftrain[i] > q3 + step)
        ].index.tolist()
        outlier_list.append(list_out)
    outliers = list(itertools.chain.from_iterable(outlier_list))
    uniq_outlier = list(set(outliers))
    dup_outliers = list(set([x for x in outliers if outliers.count(x) > 1]))
    print('uniq outliers:', len(uniq_outlier))
    print('dup outliers (w >=2 kolumnach):', len(dup_outliers))
    # drop po etykietach indeksu (po split indeks nie jest 0..n-1)
    real_data = dftrain.drop(index=dup_outliers).reset_index(drop=True)
    print('kształt po usunięciu dup-outlierów:', real_data.shape)
    return real_data, uniq_outlier, dup_outliers
```

# 1. ToxinsbaseClass — target: `Toxicity`

Toksyny w jeziorach. 23 kolumny, klasy toksyczności 1/2/3.

```python
df = pd.read_csv(DATA + 'ToxinsbaseClass.csv')
df.columns = df.columns.str.strip()
df = df.iloc[:, 0:23]   # właściwa liczba kolumn
print('kształt pełny:', df.shape)

test_size = 0.30
dftrain, dftest = train_test_split(df, test_size=test_size, random_state=42, stratify=df['Toxicity'])
print('train:', dftrain.shape, '| test:', dftest.shape)
dftrain.head()
```

```python
dftrain.info()
```

```python
dftrain.describe().T.head(12)
```

```python
print('klasy Toxicity w train:')
print(dftrain['Toxicity'].value_counts().sort_index())
dftrain.columns
```

### Korelacja (Toxins)

```python
plt.figure(figsize=(12, 10))
sns.heatmap(dftrain.corr(numeric_only=True), fmt='.2f', cmap='YlGnBu', annot=False)
plt.title('Toxins — macierz korelacji (train)')
plt.tight_layout()
plt.show()
```

**Podpis.** Heatmapa Pearsona dla 23 zmiennych zbioru toksyn.

**Wniosek.** Większość korelacji jest słaba/umiarkowana — mało silnych par. Silniejsze powiązania widać głównie między toksynami MC_* oraz wskaźnikami TDI/TTC, co ma sens chemiczny. Do modelu nie wystarczy sama korelacja między X — potrzebujemy związku z `Toxicity`.

### Rozkłady wybranych cech vs Toxicity

Bez `size=8` (wskazówka ze slajdu). Biorę kilka cech środowiskowych i toksynowych zamiast pętli po wszystkich 22.

```python
cechy_tox = ['MaximumDepth', 'SecchiDepth', 'T_Surf', 'TP', 'TN', 'Chlorophyl_a', 'TTC', 'TDI']
for i in cechy_tox:
    g = sns.FacetGrid(dftrain, hue='Toxicity', height=3.5, aspect=1.3)
    g.map(sns.histplot, i, kde=True, element='step', stat='density', common_norm=False)
    g.add_legend(title='Toxicity')
    g.fig.suptitle(f'Toxins — rozkład {i} wg Toxicity', y=1.02)
    plt.show()
    print(f'Podpis: rozkład {i} w klasach toksyczności.')
```

**Wniosek (rozkłady).** Klasy toksyczności nakładają się na wielu cechach, ale wskaźniki typu `TTC`/`TDI` i chlorofil bardziej rozsuwają grupy niż sama głębokość. To zapowiedź, że test χ² wskaże głównie zmienne związane z toksynami/żyznością, nie lokalizację.

### Test Chi-kwadrat (α = 0.25)

```python
ALPHA = 0.25
important_tox, discarded_tox, rank_tox = run_chi_selection(dftrain, 'Toxicity', alpha=ALPHA)
print('\n=== WAŻNE ===')
print(important_tox)
print('\n=== ODRZUCONE ===')
print(discarded_tox)
rank_tox
```

```python
plt.figure(figsize=(9, 6))
plot_df = rank_tox.sort_values('p')
colors = ['seagreen' if w else 'lightcoral' for w in plot_df['wazna']]
plt.barh(plot_df['cecha'], plot_df['p'], color=colors)
plt.axvline(ALPHA, color='black', linestyle='--', label=f'α = {ALPHA}')
plt.xlabel('p-value')
plt.title('Toxins — p-value χ² względem Toxicity')
plt.legend()
plt.tight_layout()
plt.show()
print('Podpis: zielone = IMPORTANT (p < 0.25), czerwone = odrzucone.')
```

**Wniosek (χ² Toxins).** Cechy z p < 0.25 zostają w modelu; reszta odpada jako niezależna od `Toxicity` przy tym progu. α=0.25 jest celowo luźne — w danych środowiskowych nie chcemy zbyt agresywnie wycinać sygnału.

### Outliery IQR + duplikaty outlierów

```python
real_tox, uniq_tox, dup_tox = iqr_outlier_cleanup(dftrain)
print('duplikaty wierszy w train:', int(dftrain.duplicated().sum()))
```

**Wniosek.** Usuwamy tylko wiersze będące outlierem w **≥2** kolumnach naraz (jak w slajdzie). Pojedyncze skrajności zostają — przy małych n środowiskowych nie wolno wycinać zbyt agresywnie.
