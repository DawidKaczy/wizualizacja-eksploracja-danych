## Wartości odstające (metoda IQR)

Wskazówka: wybrać 2 zmienne najbardziej podejrzane o wartości odstające.

Najpierw odsiewam kolumny, w których IQR = 0 (ganek, piwnica dodatkowa, basen, `MiscVal`). Tam prawie same zera, więc każda wartość dodatnia wypada jako outlier — to artefakt metody, a nie skrajny dom. Z pozostałych biorę:

- `LotArea` — największa skośność, maksimum wielokrotnie powyżej górnego odcięcia,
- `SalePrice` — zmienna celu, odpowiednik `Purchase` z Black Friday, wyraźny górny ogon.

```python
def iqr_report(series):
    p0 = series.min()
    p100 = series.max()
    q1 = series.quantile(0.25)
    q2 = series.quantile(0.5)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lc = q1 - 1.5 * iqr
    uc = q3 + 1.5 * iqr
    print(f'{series.name}: p0={p0:.1f}, p100={p100:.1f}, q1={q1:.1f}, q2={q2:.1f}, q3={q3:.1f}, iqr={iqr:.1f}')
    print(f'  lc={lc:.1f}, uc={uc:.1f}')
    print('  outliery na dole:', bool(lc > p0), '| outliery na górze:', bool(uc < p100))
    print('  liczba poza przedziałem:', int(((series < lc) | (series > uc)).sum()))
    return lc, uc

podejrzane = []
for col in df.select_dtypes(include='number').columns:
    s = df[col].dropna()
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    iqr = q3 - q1
    if iqr == 0:
        continue
    lc, uc = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    n = int(((s < lc) | (s > uc)).sum())
    podejrzane.append((col, n, round(s.skew(), 2), round(s.max() / uc, 2) if uc else np.nan))

ranking = pd.DataFrame(podejrzane, columns=['zmienna', 'outliery', 'skosnosc', 'max/uc'])
ranking.sort_values(['skosnosc', 'outliery'], ascending=False).head(8)
```

```python
lc_lot, uc_lot = iqr_report(df['LotArea'])
df['LotArea'].plot(kind='box')
plt.title('LotArea przed przycięciem')
plt.show()

lc_price, uc_price = iqr_report(df['SalePrice'])
df['SalePrice'].plot(kind='box')
plt.title('SalePrice przed przycięciem')
plt.show()
```

```python
df['LotArea'] = df['LotArea'].clip(lower=lc_lot, upper=uc_lot)
df['SalePrice'] = df['SalePrice'].clip(upper=uc_price)

df['LotArea'].plot(kind='box')
plt.title('LotArea po przycięciu')
plt.show()

df['SalePrice'].plot(kind='box')
plt.title('SalePrice po przycięciu')
plt.show()

print('LotArea po clip:', df['LotArea'].min(), '–', df['LotArea'].max())
print('SalePrice po clip:', df['SalePrice'].min(), '–', df['SalePrice'].max())
```

**Teza.** `LotArea` ma outliery po obu stronach, `SalePrice` tylko od góry (dolne odcięcie jest poniżej minimum). Przycięcie (`clip`) zostawia wszystkie domy w zbiorze, ale wciąga skrajne działki i najdroższe sprzedaże do granicy IQR. Po zabiegu wąsy wykresu pudełkowego kończą się na odcięciu — ogon cenowy nie jest już kilka razy dłuższy od reszty rozkładu.

## Brakujące wartości

Reguła z zajęć: braki powyżej 50% — kolumnę wycofujemy. Poniżej 50% — imputacja. `float` uzupełniam średnią. Kolumn `int` z brakami w tym pliku nie ma; gdyby były, wstawiłbym wartość `top` z `describe(include='all')` (najczęstsza wartość, czyli moda). Zmienne kategoryczne (`str`) uzupełniam modą, tak jak `Product_Category_2` w Black Friday.

```python
braki = pd.DataFrame({
    'braki': df.isna().sum(),
    'procent': (df.isna().mean() * 100).round(2)
})
braki = braki[braki['braki'] > 0].sort_values('procent', ascending=False)
braki
```

```python
do_usuniecia = braki.index[braki['procent'] > 50].tolist()
print('usuwam (>50% braków):', do_usuniecia)
df = df.drop(columns=do_usuniecia)

for col in df.columns[df.isna().any()]:
    if pd.api.types.is_float_dtype(df[col]):
        wartosc = df[col].mean()
        sposob = 'mean'
    else:
        # int albo kategoria: top z describe(include='all'), czyli moda
        wartosc = df[col].mode().iloc[0]
        sposob = 'mode/top'
    print(f'{col}: {sposob} = {wartosc}')
    df[col] = df[col].fillna(wartosc)

print('braki po leczeniu:', int(df.isna().sum().sum()))
print('kształt:', df.shape)
```

**Teza.** Pięć kolumn (`PoolQC`, `MiscFeature`, `Alley`, `Fence`, `MasVnrType`) jest pustych w ponad połowie wierszy — imputacja wstawiłaby jedną etykietę większości obserwacji, więc je wycofuję. Resztę da się uzupełnić bez wyrzucania domów: `LotFrontage`, `GarageYrBlt` i `MasVnrArea` średnią, cechy garażu, piwnicy, kominka i `Electrical` modą. Po leczeniu w ramce nie ma już `NaN`.

### Liczbowe i liczbowe

### Liczbowe i kategoryczne

Dla ceny biorę średnią albo medianę w grupie, nie sumę. Suma `SalePrice` premiowałaby dzielnice z większą liczbą transakcji.

```python
df.groupby('OverallQual')['SalePrice'].mean().plot(kind='bar')
plt.title('Średnia cena wg OverallQual')
plt.ylabel('SalePrice')
plt.show()

df.groupby('YearBuilt')['SalePrice'].mean().plot(kind='line')
plt.title('Średnia cena wg roku budowy')
plt.ylabel('SalePrice')
plt.show()

df.groupby('Neighborhood')['SalePrice'].median().sort_values().plot(kind='barh', figsize=(8, 7))
plt.title('Mediana ceny wg dzielnicy')
plt.xlabel('SalePrice')
plt.show()

sns.boxplot(x='KitchenQual', y='SalePrice', data=df, order=['Fa', 'TA', 'Gd', 'Ex'])
plt.title('Cena wg jakości kuchni')
plt.show()
```

**Teza.** Jakość jest wyceniana schodkowo: od oceny 5 do 10 średnia cena rośnie niemal przy każdym stopniu, a kuchnia `Ex` leży wyraźnie wyżej niż `TA`/`Fa`. Nowsze roczniki są droższe, choć linia roku budowy skacze — pojedyncze lata mają mało domów. Dzielnica rozsuwa mediany mniej więcej trzykrotnie (`MeadowV` i `IDOTRR` kontra `NoRidge` i `NridgHt`).
