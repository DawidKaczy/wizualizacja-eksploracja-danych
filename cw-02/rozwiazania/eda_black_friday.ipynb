## Duplikaty

```python
# pelne wiersze
print('duplikaty wierszy:', df.duplicated().sum())

# User_ID sie powtarza (wiele zakupow tego samego usera) - tylko podglad, bez inplace
df.drop_duplicates(subset='User_ID', inplace=False).head()
```

## Wartosci odstajace (Purchase) - metoda IQR

```python
p0 = df.Purchase.min()
p100 = df.Purchase.max()
q1 = df.Purchase.quantile(0.25)
q2 = df.Purchase.quantile(0.5)
q3 = df.Purchase.quantile(0.75)
iqr = q3 - q1

lc = q1 - 1.5 * iqr
uc = q3 + 1.5 * iqr

print('p0 =', p0, ', p100 =', p100, ', lc =', lc, ', uc =', uc)
print('outliery dol:', lc > p0)
print('outliery gora:', uc < p100)

df.Purchase.plot(kind='box')
plt.title('Purchase - przed przycieciem')
plt.show()
print('Boxplot: widać outliery powyżej górnego odcięcia (uc).')
```

```python
# przycinanie wartosci powyzej uc
# (bez inplace - w nowszym pandas clip na kolumnie potrafi wywalic)
df['Purchase'] = df['Purchase'].clip(upper=uc)

df.Purchase.plot(kind='box')
plt.title('Purchase - po przycieciu')
plt.show()
print('Po clip: wartości Purchase mieszczą się w zakresie, outliery ogarnięte.')
```

## Brakujace wartosci

```python
braki = df.isna().sum() / df.shape[0]
braki
```

```python
# Product_Category_2 ~31% braków -> imputacja moda
moda = df.Product_Category_2.mode()[0]
print('moda PC2:', moda)
df['Product_Category_2'] = df['Product_Category_2'].fillna(moda)

# Product_Category_3 ~70% braków -> wyrzucamy kolumne
df = df.dropna(axis=1)

print(df.isna().sum())
df.dtypes
```
